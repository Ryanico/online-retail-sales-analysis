```python
def classify_transaction(row):
    if row["UnitPrice"] < 0:
        return "Bad Debt"
    elif row["InvoiceNo"].startswith("C"):
        return "Cancellation/Return"
    elif row["Quantity"] < 0:
        return "Adjustment"
    elif row["UnitPrice"] == 0:
        return "Non-Sale"
    else:
        return "Sale"

clean_df["TransactionType"] = clean_df.apply(classify_transaction, axis=1)
```

# Final Business Findings and Recommendations

## 1. Executive Summary

This project analyzes an online retailer's sales transactions from December 2010 to December 2011. Python and pandas were used to clean and analyze the data, identify sales trends, evaluate product and customer performance, and develop actionable business recommendations.

The analysis recorded approximately **£10.64 million in sales revenue** across 19,960 orders. The findings reveal a strong dependence on the UK market, valuable repeat customers, differences between product sales volume and revenue, and significant variation in order values.

## 2. Key Business Findings

### Overall Sales Performance

- Total revenue: £10,642,110.80
- Total units sold: 5,572,420
- Total orders: 19,960
- Average order value: £533.17
- Median order value: £303.30

The mean order value is substantially higher than the median, indicating a right-skewed distribution in which some unusually large orders raise the average.

### Monthly Sales Trends

- November 2011 recorded the highest monthly revenue at approximately £1.50 million.
- October generated approximately £1.15 million, while September generated approximately £1.06 million.
- Revenue increased considerably between September and November.
- February 2011 was the weakest complete month by revenue.

December 2011 contains data only through December 9, so its revenue should not be compared directly with complete months.

### Product Performance

- *REGENCY CAKESTAND 3 TIER* was the highest-revenue product, generating £174,156.54.
- *PAPER CRAFT, LITTLE BIRDIE* generated £168,469.60 and led product sales volume with 80,995 units.
- *WORLD WAR 2 GLIDERS ASSTD DESIGNS* sold 54,951 units but generated only £13,814.01.

These results demonstrate that high sales volume does not necessarily translate into high revenue. Both volume and revenue should be considered when making inventory and product decisions.

### Geographic Performance

- The UK generated approximately £9.00 million, representing 84.59% of total revenue.
- The Netherlands generated £285,446.34 across 94 orders.
- EIRE generated £283,140.52, while Germany generated £228,678.40.

The retailer relies heavily on the UK market. International markets may offer growth opportunities, but their potential should be assessed using order volume, customer activity, repeat purchasing, and profitability rather than revenue alone.

### Customer Performance

- 4,338 customers had identifiable customer IDs.
- 2,845 customers (65.6%) placed multiple orders.
- Repeat customers generated approximately £8.27 million, or 93.1% of revenue attributable to identifiable customers.
- One-time customers generated approximately £614,000.
- The top 10 customers by revenue contributed approximately £1.54 million.

Repeat customers represent an important source of revenue. However, the analysis covers a limited period, so the repeat-customer share alone does not establish long-term customer retention.

Approximately £1.76 million in sales revenue could not be attributed to identifiable customers because customer IDs were missing.

### Order Behavior

- The average order value was £533.17, compared with a median of £303.30.
- The 10 highest-revenue orders contributed £508,883.38, or 4.78% of total revenue.
- The largest order generated £168,469.60 from 80,995 units of *PAPER CRAFT, LITTLE BIRDIE*.

This large order illustrates how bulk purchases can influence sales performance. Unusually large transactions should be investigated before being treated as representative of typical demand.

### Time and Purchasing Patterns

- Thursday generated the highest total weekday revenue at approximately £2.20 million.
- Tuesday recorded the highest average revenue per recorded sales date at £41,840.39.
- 10:00 generated the highest total hourly revenue at approximately £1.44 million.
- 12:00 recorded the highest hourly order count, with 3,220 orders.
- 07:00 had the highest average order value at £1,071.01, based on only 29 orders.
- December 9, 2011, generated £200,900.98, of which one bulk order contributed approximately 83.9%.

These findings indicate that revenue, order volume, and average order value peak at different times. Saturday had no recorded sales, and the reason is unknown.

## 3. Business Recommendations

### 1. Strengthen Customer Retention

Develop targeted loyalty initiatives, personalized product recommendations, and repeat-purchase campaigns. Analyze purchasing frequency and time between orders to identify customers who may be at risk of becoming inactive.

### 2. Improve Product and Inventory Decisions

Use both revenue and unit sales to guide inventory planning. Maintain appropriate stock levels for high-performing products while reviewing the commercial value of high-volume, low-revenue items.

### 3. Investigate International Growth Opportunities

Assess selected international markets, including the Netherlands, EIRE, and Germany, for sustainable growth. Compare order frequency, average order value, shipping costs, and profitability before allocating additional marketing resources.

### 4. Monitor Large Orders and Revenue Concentration

Flag unusually large orders for review and distinguish bulk purchases from typical customer demand. This can improve forecasting and help management understand whether revenue spikes are repeatable.

### 5. Use Time-Based Insights Carefully

Investigate why Tuesday has relatively high average daily revenue and why midday hours generate substantial order volume. Validate these patterns across comparable periods before changing staffing, promotions, or operational schedules.

### 6. Improve Customer Data Quality

Investigate missing customer IDs and establish processes to improve customer identification. Better customer data would strengthen segmentation, repeat-purchase analysis, and customer lifetime value estimates.

## 4. Limitations

- Customer-level analysis excludes transactions without identifiable customer IDs.
- The dataset covers approximately one year, limiting conclusions about long-term trends and seasonality.
- December 2011 is incomplete.
- Some average order values are influenced by unusually large transactions or small order counts.
- Revenue does not equal profit because product costs, shipping costs, operating expenses, and margins are not available.
- Observed patterns show associations, not proof of what caused changes in sales.

## 5. Conclusion

The analysis shows that the retailer's performance is supported by a dominant UK market, repeat-customer spending, and several high-performing products. However, sales are unevenly distributed across products, customers, dates, and hours.

The strongest next steps are to improve customer data quality, strengthen repeat purchasing, refine inventory decisions, and investigate international growth opportunities using profitability and customer behavior.

The findings provide a foundation for the Power BI dashboard, where the most important KPIs, trends, and business recommendations can be communicated visually.
